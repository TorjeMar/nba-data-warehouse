```python
import json
```

```python
def prettyprint(x):
    print(json.dumps(x, indent=4))
```

```python
def read_jsonl(path: str) -> list[dict]:
    with open(path, 'r') as f:
        return list(map(json.loads, f))
```

```python
data = read_jsonl('data/box_scores.jsonl')
```
