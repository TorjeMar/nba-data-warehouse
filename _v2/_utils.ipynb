```python
def create_notebooks(notebooks: list[str], template: dict, directory: str):
    disk.makedirs(directory, exist_ok=True)
    
    for idx, notebook in enumerate(notebooks, start=1):
        path = disk.joinpath(directory, f'{idx:03d}_{notebook}.ipynb')
        if disk.isfile(path):
            print(f'Skipping {path} ')
            continue
        disk.write_json(path, template)
```
